- `size` - uy maydoni (kv.m)

```python
import numpy as np
```

```python
ind=df[df['price']=='Договорная'].index
df['price'].replace('Договорная',np.nan,inplace=True)
```

```python
df['price']=pd.to_numeric(df['price'])
```

```python
ind=df[df['size']=='Площадьземли:1сот'].index
df['size'].replace('Площадьземли:1сот',np.nan,inplace=True)
```

```python
df['size']=pd.to_numeric(df['size'])
```

```python
df.info()
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestRegressor

# 1. Define feature groups
cat = ['location', 'district']
num = ['rooms', 'level', 'size', 'max_levels']

# 2. Column transformer pipeline (corrected)
preprocessor = ColumnTransformer([
    ('num', Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ]), num),

    ('cat', OneHotEncoder(handle_unknown='ignore'), cat)
])

# 3. Split known and unknown prices
df_notnan = df[df['price'].notna()]
df_nan = df[df['price'].isna()]

X_notnan = df_notnan.drop('price', axis=1)
Y_notnan = df_notnan['price']

X_nan = df_nan.drop('price', axis=1)

# 4. Create imputation model pipeline
imputer_model = Pipeline([
    ('preprocessing', preprocessor),
    ('regressor', RandomForestRegressor(random_state=42))
])

# 5. Fit model and fill missing prices
imputer_model.fit(X_notnan, Y_notnan)
df.loc[df['price'].isna(), 'price'] = imputer_model.predict(X_nan)
```

```python
df['size']=df['size'].fillna(df['size'].mean())
```

```python
df.head() #how deos this model owrks for one new data, do i need to pipeline that one, or directly
```

```python
X=df.drop('price',axis=1)
Y=df['price']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test=train_test_split(X,Y,random_state=42,test_size=0.2)
```

```python
numericpip=Pipeline([
    ('scaler', StandardScaler())
])
categoricalpip=Pipeline([
  ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

fullpipeline=ColumnTransformer([
    ('categorical', categoricalpip, cat),
    ('numerical', numericpip, num)
])
```

```python
X_train_prepared = fullpipeline.fit_transform(X_train)
```

```python
X_test_prepared = fullpipeline.transform(X_test)
```

```python
RF_model=RandomForestRegressor()
RF_model.fit(X_train_prepared,Y_train)
Y_predicted = RF_model.predict(X_test_prepared)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

RF_mse=mean_squared_error(Y_test, Y_predicted)
print(np.sqrt(RF_mse))
```
